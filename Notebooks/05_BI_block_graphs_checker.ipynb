# Block Graphs BI Vanishing Check

This notebook will first check whether the Bond Information is equal to $0$ for all bond factorisations of the joint, for some common connected block graph families, using the analytic Gaussian bond information. We do this by finding the BI's for common graph families up to $6$ vertices, then doing a check for all graph families at once. This notebook supports Theorem $5.2.2$ in the report. Afterwards, it will check that this holds for all block graphs up to $6$ vertices.

```python
import networkx as nx
import numpy as np
from networkx.generators.atlas import graph_atlas_g
from Modules.gaussian_bond_information import *
from Modules.bond_finders import *
from Modules.lattice_machinery import *
```

```python
# Complete graphs (is just streitberg so will hold)

# We will separate the value of BI for factorisations of the joint and the fully dependent joint
fact_informations = []
full_informations = []

for i in range(1, 7):
    K = nx.complete_graph(i)
    bis, _ = factorised_bi(K, walk_bond_finder, 0.99)
    
    fact_informations.append(list(bis.values())[:-1])
    full_informations.append(list(bis.values())[-1])
```

```python
# Path graphs
informations_path = []
full_informations_path = []

for i in range(1, 7):
    K = nx.path_graph(i)
    bis_path, _ = factorised_bi(K, walk_bond_finder, 0.99)

    informations_path.append(list(bis_path.values())[:-1])
    full_informations_path.append(list(bis_path.values())[-1])
```

```python
# Star graphs
informations_star = []
full_informations_star = []

for i in range(1, 7):
    K = nx.star_graph(i-1)
    bis_star, _ = factorised_bi(K, walk_bond_finder, 0.99)

    informations_star.append(list(bis_star.values())[:-1])
    full_informations_star.append(list(bis_star.values())[-1])
```

```python
# Check that whether all bond factorisations give BI = 0
print("Checking whether Bond Information vanishes on all bond factorisations")
print("-"*69)
print("Complete graphs:", all(np.allclose(vals, 0) for vals in fact_informations))
print("Path graphs:", all(np.allclose(vals, 0) for vals in informations_path))
print("Star graphs:", all(np.allclose(vals, 0) for vals in informations_star))
```

```python
# Check that whether fully dependent joint distributions give BI != 0
print("Checking whether Bond Information does not vanish on fully dependent joint distribution")
print("-"*87)
print("Complete graphs:", not all(np.allclose(vals, 0) for vals in full_informations))
print("Path graphs:", not all(np.allclose(vals, 0) for vals in full_informations_path))
print("Star graphs:", not all(np.allclose(vals, 0) for vals in full_informations_star))
```

Thus, on complete, star and path graphs up to $6$ vertices, the analytic Gaussian Bond Information vanishes when the joint distribution factorises according to a bond, and does not vanish when the joint is fully dependent. We now will do a more exhaustive sweep of this verification for all connected block graphs up to $6$ vertices.

```python
all_graphs = graph_atlas_g()
graphs = [
    G for G in all_graphs
    if 1 <= G.number_of_nodes() <= 6
]
```

```python
# We will separate the value of BI for factorisations of the joint and the fully dependent joint
fact_informations = []
full_informations = []

for G in graphs:
    if is_block_graph(G) == True:
        bis, _ = factorised_bi(G, walk_bond_finder, 0.99)
    
        fact_informations.append(list(bis.values())[:-1])
        full_informations.append(list(bis.values())[-1])

# Check that whether all bond factorisations give BI = 0
print("Checking whether Bond Information vanishes on all bond factorisations, for all block graphs up to 6 vertices")
print("-"*69)
print(all(np.allclose(vals, 0) for vals in fact_informations))

print("")

# Check that whether all fully-dependent joints give BI != 0
print("Checking whether Bond Information does not vanish on fully-dependent joint, for all block graphs up to 6 vertices")
print("-"*69)
print(not all(np.allclose(vals, 0) for vals in full_informations))
```
