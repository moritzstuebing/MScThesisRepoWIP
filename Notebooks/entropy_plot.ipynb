# Entropy Plot

This will be a very quick and simple notebook used to produce the entropy plot given in Example $3.1$.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set(style="whitegrid")
```

```python
p = np.linspace(0.01, 0.99, 10000)
y = (- p * np.log(p) - (1 - p) * np.log(1 - p)) / np.log(2)
```

```python
plt.figure(figsize=(14, 8))
plt.tick_params(axis='both', labelsize=20)
plt.plot(p, y)
plt.xlabel(r"$p$", fontsize=20)
plt.ylabel(r"$H(X)$", fontsize=20)
plt.tight_layout()
plt.savefig("../Plots/entropy_plot.pdf")
plt.show()
```
