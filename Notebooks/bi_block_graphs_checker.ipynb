# Block Graphs BI Vanishing Check

This notebook will check whether the Bond Information is equal to $0$ for all bond factorisations of the joint, for some common block graph families.

```python
import networkx as nx
import numpy as np
from Modules.gaussian_BI import covariance_matrix_bond, factorised_bi, bond_nonbond_list_maker, analytic_gaussian_bi, restrict
from Modules.brute_force_bond_finder import brute_force_bond_finder
```

```python
# Complete graphs (is just streitberg so will hold)

# We will separate the value of BI for factorisations of the joint and the fully dependent joint
fact_informations = []
full_informations = []

for i in range(1, 7):
    print(i)
    K = nx.complete_graph(i)
    bis, _ = factorised_bi(K, brute_force_bond_finder, 0.99)
    
    fact_informations.append(list(bis.values())[:-1])
    full_informations.append(list(bis.values())[-1])
```

```python
# Path graphs
informations_path = []
full_informations_path = []

for i in range(1, 7):
    K = nx.path_graph(i)
    bis_path, _ = factorised_bi(K, brute_force_bond_finder, 0.99)

    informations_path.append(list(bis_path.values())[:-1])
    full_informations_path.append(list(bis_path.values())[-1])
```

```python
# Stars graphs
informations_star = []
full_informations_star = []

for i in range(1, 7):
    K = nx.star_graph(i)
    bis_star, _ = factorised_bi(K, brute_force_bond_finder, 0.99)

    informations_star.append(list(bis_star.values())[:-1])
    full_informations_star.append(list(bis_star.values())[-1])
```

```python
# Check that whether all bond factorisations give BI = 0
# Each graph size has a different number of bonds, so these are ragged lists of lists -
# check each graph's BI values separately rather than stacking into one array
print("Complete graphs:", all(np.allclose(vals, 0) for vals in fact_informations))
print("Path graphs:", all(np.allclose(vals, 0) for vals in informations_path))
print("Star graphs:", all(np.allclose(vals, 0) for vals in informations_star))
```
