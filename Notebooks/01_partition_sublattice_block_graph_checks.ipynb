# Partition Sublattice / Block Graphs Checks

This notebook will serve as a verification of Theorem $4.0.32$. We first check whether the bond lattices of various common graph types up to $8$ vertices are sublattices of the partition lattice. Since the bond join is the same as the partition join, we only need to check that the bond lattice is closed under the partition meet. This is precisely what the function partition_sublattice_checker() does. We present the findings in Table $4.1$ in the report. Secondly, we check whether the block graph / partition sublattice equivalence holds for all connected graphs up to $7$ vertices. 

```python
import networkx as nx
from networkx.generators.atlas import graph_atlas_g
from Modules.bond_finders import *
from Modules.lattice_machinery import *
```

## 1. Complete Graphs

```python
# Loop over number of vertices d
for i in range(1, 9):

    # Form graph with specified number of vertices
    graph = nx.complete_graph(i)

    # Check whether sublattice of partition lattice
    verdict = partition_sublattice_checker(graph, walk_bond_finder)

    print(f"{i}-vertex complete graph's bond lattice is a sublattice of the partition lattice: {verdict}")
```

## 2. Path Graphs

```python
# Loop over number of vertices d
for i in range(1, 9):

    # Form graph with specified number of vertices
    graph = nx.path_graph(i)

    # Check whether sublattice of partition lattice
    verdict = partition_sublattice_checker(graph, walk_bond_finder)
    print(f"{i}-vertex path graph's bond lattice is a sublattice of the partition lattice: {verdict}")
```

## 3. Star Graphs

```python
# Loop over number of vertices d
for i in range(0, 8):

    # Form graph with specified number of vertices
    graph = nx.star_graph(i)

    # Check whether sublattice of partition lattice
    verdict = partition_sublattice_checker(graph, walk_bond_finder)
    print(f"{i+1}-vertex star graph's bond lattice is a sublattice of the partition lattice: {verdict}")
```

## 4. Wheel Graphs

```python
# Loop over number of vertices d
for i in range(1, 9):

    # Form graph with specified number of vertices
    graph = nx.wheel_graph(i)

    # Check whether sublattice of partition lattice
    verdict = partition_sublattice_checker(graph, walk_bond_finder)

    print(f"{i}-vertex wheel graph's bond lattice is a sublattice of the partition lattice: {verdict}")
```

## 5. Cycle Graphs

```python
# List to store sublattice verdicts
verdicts = []

# Loop over number of vertices d
for i in range(1, 9):

    # Form graph with specified number of vertices
    graph = nx.cycle_graph(i)

    # Check whether sublattice of partition lattice
    verdict = partition_sublattice_checker(graph, walk_bond_finder)

    print(f"{i}-vertex cycle graph's bond lattice is a sublattice of the partition lattice: {verdict}")
```

## 6. All connected graphs up to 7 vertices

```python
# Retrive the graph atlas generator
all_graphs = graph_atlas_g()

# Keep only connected graphs between 1 and 7 vertices
graphs = [
    G for G in all_graphs
    if 1 <= G.number_of_nodes() <= 7 and nx.is_connected(G)
]
```

```python
verdicts = []

# Loop over all connected graphs with 1 to 7 vertices
for G in graphs:

    # Check the property equivalence holds
    verdict = (partition_sublattice_checker(G, walk_bond_finder) == is_block_graph(G))
    verdicts.append(verdict)

    # If equivalence does not hold for some graph, make note of this
    if verdict == False:
        print(f"Failed for {G}")

# If equivalence holds for all graphs, print statement
if all(verdicts) == True:
    print("Block graph / partition sublattice check holds for all connected graphs up to 7 vertices")
```
