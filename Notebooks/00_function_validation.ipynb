# 1. Brute Force Algorithm

Firstly, we will check that our brute force bond-finding algorithm works correctly. We will first do this by checking that the number of bonds it outputs is as expected. We test this with complete graphs where we expect outputs in accordance with the Bell numbers, and some tree graphs where we expect the nubmer of bonds to be $2^{N-1}$. We will test this for up to $7$ vertices.

```python
# Complete graphs

# Theoretical lattice sizes
bell_numbers = [1, 2, 5, 15, 52, 203, 877]

for i in range(1, 8):

    # Graph and bonds generation
    graph = nx.complete_graph(i)
    bonds = brute_force_bond_finder(graph)

    # Checks
    print(f"Number of vertices: {i}. Number of bonds: {len(bonds)}. Same as Bell number: {len(bonds) == bell_numbers[i-1]}")
```

```python
# Path graphs

for d in range(1, 8):

    # Theoretical lattice size
    theoretical_size = 2 ** (d-1)

    # Graph and bonds generation
    graph = nx.path_graph(d)
    bonds = brute_force_bond_finder(graph)

    # Checks
    print(f"Number of vertices: {d}. Number of bonds: {len(bonds)}. Same as theoretical lattice size: {len(bonds) == theoretical_size}")
```

```python
# Star graphs

for d in range(1, 8):

    # Theoretical lattice size
    theoretical_size = 2 ** (d-1)

    # Graph and bonds generation
    graph = nx.star_graph(d-1) # argument is the number of outer nodes
    bonds = brute_force_bond_finder(graph)

    # Checks
    print(f"Number of vertices: {d}. Number of bonds: {len(bonds)}. Same as theoretical lattice size: {len(bonds) == theoretical_size}")
```

```python
# Cycle graphs

for d in range(1, 8):

    # Theoretical lattice size
    theoretical_size = 2 ** d - d

    # Graph and bonds generation
    graph = nx.cycle_graph(d) # argument is the number of outer nodes
    bonds = brute_force_bond_finder(graph)

    # Checks
    print(f"Number of vertices: {d}. Number of bonds: {len(bonds)}. Same as theoretical lattice size: {len(bonds) == theoretical_size}")
```

## 2. Other Bond-Finders

Now that we can be confident that our brute force algorithm is correctly producing all of the bonds of a graph, we can use it to validate that our other two algorithms are also working correctly. Since all algorithms use the same bond sorting at the end, they should produce the exact same bonds in the exact same positions, so we can directly compare the lists of bonds that the algorithms produce. We use all of our usual graph families to check this again. We keep the function below in this notebook rather than in a module since it will only be used here.

```python
def algorithm_checks(brute, edge, walk, family, d):

    """
        Function to check whether the outputs of all algorithms are the same for a given graph family
    """

    if family == nx.star_graph: # the star graph generator has one central vertex and d outer vertices
        for i in range(0, d-1):

                # Generate graph
                G = family(i)

                # Retrieve bonds
                brute_bonds = brute(G)
                edge_bonds = edge(G)
                walk_bonds = walk(G)

                # Check equalities
                verdict_1 = brute_bonds == edge_bonds
                verdict_2 = brute_bonds == walk_bonds
                verdict_3 = edge_bonds == walk_bonds

                # Print results
                if verdict_1 and verdict_2 and verdict_3 is True:
                    print(f"For {i+1} vertices, all algorithms produce the same bonds.")
        
                else:
                    print(f"Different bond outputs for number of vertices {i+1}")
        
    else:
        for i in range(1, d):

            # Generate graph
            G = family(i)

            # Retrieve bonds
            brute_bonds = brute(G)
            edge_bonds = edge(G)
            walk_bonds = walk(G)

            # Check equalities
            verdict_1 = brute_bonds == edge_bonds
            verdict_2 = brute_bonds == walk_bonds
            verdict_3 = edge_bonds == walk_bonds

            # Print results
            if verdict_1 and verdict_2 and verdict_3 is True:
                print(f"For {i} vertices, all algorithms produce the same bonds.")

            else:
                print(f"Different bond outputs for number of vertices {i}")
```

```python
print("Complete Graphs:")
print("-"*54)
algorithm_checks(
    brute_force_bond_finder,
    edge_bond_finder,
    walk_bond_finder,
    nx.complete_graph,
    8)
```

```python
print("Path Graphs:")
print("-"*54)
algorithm_checks(
    brute_force_bond_finder,
    edge_bond_finder,
    walk_bond_finder,
    nx.path_graph,
    8)
```

```python
print("Star Graphs:")
print("-"*54)
algorithm_checks(
    brute_force_bond_finder,
    edge_bond_finder,
    walk_bond_finder,
    nx.star_graph,
    8)
```

```python
print("Wheel Graphs:")
print("-"*54)
algorithm_checks(
    brute_force_bond_finder,
    edge_bond_finder,
    walk_bond_finder,
    nx.wheel_graph,
    8)
```

```python
print("Cycle Graphs:")
print("-"*54)
algorithm_checks(
    brute_force_bond_finder,
    edge_bond_finder,
    walk_bond_finder,
    nx.cycle_graph,
    8)
```

For all of the checks we have done, all three algorithms produce the same bonds. This strongly supports that we truly are finding the correct bonds, since three distinct algorithms generate the same bonds, and since we can be almost certain the brute force algorithm works correctly due to the previous verification.

```python
# Retrieve all connected graphs up to 6 vertices
all_graphs = graph_atlas_g()
graphs = [
    G for G in all_graphs
    if 1 <= G.number_of_nodes() <= 6 and nx.is_connected(G)
]
```

```python
verdicts = []

for G in graphs:

    # Retrieve bonds
    brute_bonds = brute_force_bond_finder(G)
    edge_bonds = edge_bond_finder(G)
    walk_bonds = walk_bond_finder(G)
    
    # Check equalities
    verdict_1 = brute_bonds == edge_bonds
    verdict_2 = brute_bonds == walk_bonds
    verdict_3 = edge_bonds == walk_bonds
    
    verdicts.append(verdict_1 and verdict_2 and verdict_3)

# Check algorithms agree on all graphs
print(all(verdicts))
```

## 3. Moebius Coefficients

Now we will briefly validate our Moebius coefficients function to check that it is producing the correct coefficients. Firstly, we check that for all connected graphs up to $6$ vertices, our function correctly produces a Moebius matrix with ones along the diagonal, since these entries correspond to the $\mu(x, x)$ values which should equal $1$. We also check that all of the rows corresponding to $x \ne \hat{1}$ of the matrix sum to zero, which corresponds to the identity $\sum_{x \le y \le z} \mu(x, y) = 0$ for $x < z$.

```python
graphs_6 = [
    G for G in all_graphs
    if G.number_of_nodes() == 6 and nx.is_connected(G)
]
```

```python
verdicts_xx = []
verdicts_row_sums = []

for G in graphs_6:
    _, _, moebius, _ = full_inv_mu(G, walk_bond_finder)

    # Diagonal check
    diag = moebius.diagonal()
    verdict_xx = (all(diag == np.ones(moebius.shape[0])))
    verdicts_xx.append(verdict_xx)

    # Row sums check
    row_sums = np.sum(moebius, axis=1)
    row_sums = row_sums[:-1]
    verdict_rows = all(row_sums == 0)
    verdicts_row_sums.append(verdict_rows)

print(f"All diagonals of all Moebius matrices of graphs equal 1: {all(verdicts_xx)}")
print(f"All row sums of all Moebius matrices of all graphs equal zero: {all(verdicts_row_sums)}")
```
