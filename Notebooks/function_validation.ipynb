# 1. Brute Force Algorithm

Firstly, we will check that our brute force bond-finding algorithm works correctly. We will first do this by checking that the number of bonds it outputs is as expected. We test this with complete graphs where we expect outputs in accordance with the Bell numbers, and some tree graphs where we expect the nubmer of bonds to be $2^{N-1}$. We will test this for up to $7$ vertices.

```python
# Complete graphs

# Theoretical lattice sizes
bell_numbers = [1, 2, 5, 15, 52, 203, 877]

for i in range(1, 8):

    # Graph and bonds generation
    graph = nx.complete_graph(i)
    bonds = brute_force_bond_finder(graph)

    # Checks
    print(f"Number of vertices: {i}. Number of bonds: {len(bonds)}. Same as Bell number: {len(bonds) == bell_numbers[i-1]}")
```

```python
# Path graphs

for d in range(1, 8):

    # Theoretical lattice size
    theoretical_size = 2 ** (d-1)

    # Graph and bonds generation
    graph = nx.path_graph(d)
    bonds = brute_force_bond_finder(graph)

    # Checks
    print(f"Number of vertices: {d}. Number of bonds: {len(bonds)}. Same as theoretical lattice size: {len(bonds) == theoretical_size}")
```

```python
# Star graphs

for d in range(1, 8):

    # Theoretical lattice size
    theoretical_size = 2 ** (d-1)

    # Graph and bonds generation
    graph = nx.star_graph(d-1) # argument is the number of outer nodes
    bonds = brute_force_bond_finder(graph)

    # Checks
    print(f"Number of vertices: {d}. Number of bonds: {len(bonds)}. Same as theoretical lattice size: {len(bonds) == theoretical_size}")
```

```python
# Cycle graphs

for d in range(1, 8):

    # Theoretical lattice size
    theoretical_size = 2 ** d - d

    # Graph and bonds generation
    graph = nx.cycle_graph(d) # argument is the number of outer nodes
    bonds = brute_force_bond_finder(graph)

    # Checks
    print(f"Number of vertices: {d}. Number of bonds: {len(bonds)}. Same as theoretical lattice size: {len(bonds) == theoretical_size}")
```

For a few small examples, we will now check that the brute force bond finder is outputting the correct bonds. We begin with $K_3$ and $K_4$. The bonds for these should be any partition of $3$ and $4$ variables.

```python
print("3-Variable Complete Graph:")
brute_force_bond_finder(nx.complete_graph(3), nice_format_view=True)

print("")

print("4-Variable Complete Graph")
brute_force_bond_finder(nx.complete_graph(4), nice_format_view=True)
```

These are all the valid bonds. Now we check for path and star graphs.

```python
print("3-Variable Path Graph:")
brute_force_bond_finder(nx.path_graph(3), nice_format_view=True)

print("")

print("4-Variable Path Graph")
brute_force_bond_finder(nx.path_graph(4), nice_format_view=True)
```

```python
print("3-Variable Star Graph:")
brute_force_bond_finder(nx.star_graph(2), nice_format_view=True)

print("")

print("4-Variable Star Graph")
brute_force_bond_finder(nx.star_graph(3), nice_format_view=True)
```

These are all correct. Finally, cycles.

```python
print("3-Variable Cycle Graph:")
brute_force_bond_finder(nx.cycle_graph(3), nice_format_view=True)

print("")

print("4-Variable Cycle Graph")
brute_force_bond_finder(nx.cycle_graph(4), nice_format_view=True)
```

The brute force algorithm exhaustively checks every single possible partition of a graph's vertices, and bond_checker correctly checks the connectedness of every block of each partition, keeping only those partitions where every block gives a connected subgraph. Combined with the verification results above, we can conclude that the brute force algorithm correctly finds all of the bonds of a given graph.
