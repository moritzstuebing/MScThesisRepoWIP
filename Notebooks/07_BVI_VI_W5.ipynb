# $VI$ vs $BVI$ for Wheel Graph $W_5$

This notebook will compare the values of $VI$ and $BVI$ for the bond partitions $\pi = 1 \mid 245 \mid 3$ and $\sigma = 15 \mid 234$. $VI$ is computed using the usual formula, and $BVI$ using the built-in shortest-path algorithm in NetworkX. These calculations appear in Example $6.2.12$ of the report.

```python
import numpy as np
import networkx as nx
from Modules.bond_finders import *
from Modules.lattice_machinery import *
```

```python
# Generating the graph and its Hasse diagram
W_5 = nx.wheel_graph(5)
hasse = hasse_creator(W_5, walk_bond_finder, weight=True)

# Defining pi and sigma
pi = frozenset({frozenset({0}), frozenset({2}), frozenset({1, 3, 4})})
sigma = frozenset({frozenset({0, 4}), frozenset({1, 2, 3})})
pi_sigma_join = frozenset({frozenset({0, 1, 2, 3, 4})})
pi_sigma_meet = bond_meet(pi, sigma, W_5, brute_force_bond_finder)
```

```python
p_meet = partition_meet(pi, sigma)
vi = 2 * partition_entropy(p_meet) - partition_entropy(pi) - partition_entropy(sigma)
print(f"VI = {vi}")
```

```python
# Computing join cost and meet cost of pi and sigma in the bond lattice
join_cost = partition_entropy(pi) + partition_entropy(sigma) - 2 * partition_entropy(pi_sigma_join)
meet_cost = 2 * partition_entropy(pi_sigma_meet) - partition_entropy(pi) - partition_entropy(sigma)
print(f"Join cost: {join_cost}")
print(f"Meet cost: {meet_cost}")
```

```python
# Computing BVI and the path generating it
path = nx.shortest_path(hasse, pi, sigma, weight="weight")
bvi = nx.shortest_path_length(hasse, pi, sigma, weight="weight")
print(f"BVI = {bvi}")
print("Path generating BVI is:")
for bond in path:
    bond_viewer(bond)
```
