# Partition Sublattice Checks

This notebook will check whether the bond lattices of various common graph types are sublattices of the partition lattice. Since the bond join is the same as the partition join, we only need to check that the bond lattice is closed under the partition meet. This is precisely what the function partition_sublattice_checker() does. We present the findings in Table $1$ in the report.

```python
import networkx as nx
from Modules.bond_finders import brute_force_bond_finder
from Modules.lattice_machinery import partition_meet, partition_sublattice_checker
```

## 1. Complete Graphs

```python
# List to store sublattice verdicts
verdicts = []

# Loop over number of vertices d
for i in range(1, 9):

    # Form graph with specified number of vertices
    graph = nx.complete_graph(i)

    # Check whether sublattice of partition lattice
    verdict = partition_sublattice_checker(graph, brute_force_bond_finder)
    verdicts.append(verdict)

print(verdicts)
```
