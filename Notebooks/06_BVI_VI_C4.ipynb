# $BVI$ vs. $VI$ for $C_4$

This notebook will compare the value of BVI and VI for $\pi = 1|234$ and $\sigma=3|124$. VI can be computed simply using partition meet and entropies, while BVI requires us to run a shortest path algorithm on the weighted Hasse diagram. These calculations appear in Example $6.14$ in the report.

```python
import numpy as np
import networkx as nx
from itertools import combinations
from collections import deque
from math import log2
from Modules.bond_finders import brute_force_bond_finder
from Modules.lattice_machinery import hasse_creator, cover_check, partition_entropy, partition_meet
```

```python
pi = frozenset({frozenset({0}), frozenset({1, 2, 3})})
sigma = frozenset({frozenset({2}), frozenset({0, 1, 3})})
```

```python
# Forming graph, obtaining bonds, and creating weighted Hasse diagram
C_4 = nx.cycle_graph(4)
bonds_C_4 = brute_force_bond_finder(C_4)
hasse = hasse_creator(C_4, brute_force_bond_finder, weight=True)
```

```python
# Obtaining partition meet and verifying it is not in the bond lattice
part_meet = partition_meet(pi, sigma)
print(part_meet)
print(f"The partition meet of pi and sigma is a bond of C_4: {part_meet in bonds_C_4}")
```

```python
# Calculating Vi
VI = 2 * partition_entropy(part_meet) - partition_entropy(pi) - partition_entropy(sigma)
print("Variation of Information (VI):", VI)
```

```python
# Calculating BVI
bvi = nx.shortest_path_length(hasse, source=pi, target=sigma, weight="weight")
print(f"Bond Variation of Information (BVI): {bvi}")
```

```python
path = nx.shortest_path(hasse, source=pi, target=sigma, weight="weight")
print(path)
```
